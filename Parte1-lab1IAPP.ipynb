```python
# Estado inicial
saldo = 500_000      # Saldo de la cuenta
deposito = 150_000   # Monto a depositar
retiro = 80_000      # Monto a retirar
interes = 0.02       # 2% mensual

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero
saldo_final = saldo + (saldo * interes)

# Salida del estado final
print(f"Saldo final de la cuenta:${saldo_final:.2f}")
```

```python
# Estado inicial
saldo = float(input("Ingrese el saldo inicial de la cuenta:"))
deposito = float(input("\nIngrese el monto a depositar:"))
retiro = float(input("\nIngrese el monto a retirar:"))
interes = float(input("\nIngrese el interés mensual:"))

# Actualización del estado
saldo = saldo + deposito   # Se deposita dinero
saldo = saldo - retiro     # Se retira dinero
saldo_final = saldo + (saldo * interes)

# Salida del estado final
print(f"\nSaldo final de la cuenta:${saldo_final:.2f}")
```

El estado inicial representa los valores con los que arranca el programa: saldo = 500 000, deposito = 150 000, retiro = 80 000. El estado final representa cómo quedan esas variables al terminar: saldo: 570 000, interes: 0.02 y saldo_final = 581 400. En general estado significa "qué valor tiene cada variable en un momento dado".

Cada instrucción afecta a la memoria ya que los valores se van actualizando. Primero, se hace la declaración reservando un espacio a la memoria, le asigna un nombre a la variable y guarda en él un valor. Luego, se hace una reasignación, primero se evalúa la expresión de la derecha con los valores actuales, y luego el resultado se guarda en la misma variable. El valor anterior se sobreescribe y se pierde.

```python
# Lista inicial de temperaturas en °C
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

# Operaciones básicas
print("Temperaturas:", temperaturas)
temperaturas.append(26.0)
temperaturas.pop(0)
temperaturas.insert(1, 23.7)

temperaturas_ordenadas = sorted(temperaturas)
print("Ordenadas:", temperaturas_ordenadas)

temperatura_mayor = max(temperaturas)
temperatura_menor = min(temperaturas)
diferencia = temperatura_mayor - temperatura_menor
promedio = sum(temperaturas) / len(temperaturas)
print("Diferencia:", round(diferencia, 2))
print("Promedio:", round(promedio, 2))

if 23.7 in temperaturas:
    posicion = temperaturas.index(23.7)
    print("La temperatura 23.7 está en el índice:", posicion)
else:
    print("La temperatura 23.7 no está en la lista")

print("De la 2da a la 4ta posición:", temperaturas[1:4])
print("Las tres primeras:", temperaturas[:3])
```

```python
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

# 1. Nombre del primer producto
print("Primer producto:", inventario[0]["nombre"])

# 2. Agregar un nuevo producto
nuevo_producto = {"nombre": "Arroz", "precio": 2500, "stock": 0}
inventario.append(nuevo_producto)

# 3. Vender 2 unidades de Leche
for producto in inventario:
    if producto["nombre"] == "Leche":
        producto["stock"] = producto["stock"] - 2

# 4. Valor total del inventario
valor_total = 0
for producto in inventario:
    valor_total = valor_total + producto["precio"] * producto["stock"]
print("Valor total del inventario:", valor_total)

# 5. Productos con stock menor a 10
print("Productos con poco stock:")
for producto in inventario:
    if producto["stock"] < 10:
        print(producto["nombre"])

# 6 y 7. Subir precios 10% y redondear
for producto in inventario:
    producto["precio"] = producto["precio"] * 1.10
    producto["precio"] = round(producto["precio"], 2)

# 8. Buscar producto por nombre
nombre_buscado = input("Nombre del producto a buscar: ")
encontrado = False
for producto in inventario:
    if producto["nombre"] == nombre_buscado:
        print("Producto encontrado:", producto)
        encontrado = True
if encontrado == False:
    print("El producto no está en el inventario")

# 9. Eliminar productos con stock 0
inventario_nuevo = []
for producto in inventario:
    if producto["stock"] != 0:
        inventario_nuevo.append(producto)
inventario = inventario_nuevo

# 10. Ordenar por precio ascendente
def obtener_precio(producto):
    return producto["precio"]

inventario = sorted(inventario, key=obtener_precio)
print("Inventario ordenado por precio:", inventario)
```
